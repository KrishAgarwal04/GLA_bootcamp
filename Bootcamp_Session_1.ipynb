```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'student_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'attendance_percent': [85, 92, 78, 95, 88, 67, 90, 72, 81, 98],
    'assignment_score': [80, 90, 70, 95, 85, 60, 88, 75, 82, 97],
    'quiz_score': [75, 88, 72, 96, 82, 65, 91, 70, 85, 98],
    'lab_completed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'Yes']
}
```

```python
df = pd.DataFrame(data)

df['total_score'] = (df['assignment_score'] + df['quiz_score']) / 2

def assign_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

df['grade'] = df['total_score'].apply(assign_grade)

df['pass_fail'] = np.where(df['total_score'] >= 70, 'Pass', 'Fail')

display(df)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['attendance_percent'], bins=5, kde=True)
plt.title('Distribution of Attendance Percentage')
plt.xlabel('Attendance Percentage')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.75)
plt.show()

```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='assignment_score', y='quiz_score', hue='grade', data=df, s=100)
plt.title('Assignment Score vs Quiz Score')
plt.xlabel('Assignment Score')
plt.ylabel('Quiz Score')
plt.grid(True, alpha=0.75)
plt.show()

```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='student_id', y='total_score', data=df, palette='viridis')
plt.title('Total Score by Student ID')
plt.xlabel('Student ID')
plt.ylabel('Total Score')
plt.grid(axis='y', alpha=0.75)
plt.show()

```

```python
plt.figure(figsize=(7, 7))
df['lab_completed'].value_counts().plot.pie(
    autopct='%1.1f%%',
    startangle=90,
    colors=['skyblue', 'lightgreen']
)
plt.title('Proportion of Lab Completion')
plt.ylabel('')
plt.show()

```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='grade', data=df, order=['A', 'B', 'C', 'D', 'F'], palette='pastel')
plt.title('Distribution of Grades')
plt.xlabel('Grade')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.75)
plt.show()

```

```python
plt.figure(figsize=(7, 5))
sns.countplot(x='pass_fail', data=df, palette='coolwarm')
plt.title('Pass/Fail Distribution')
plt.xlabel('Status')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
